```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("cleaned_new_df.csv")
df['Year']=df['Year'].astype(int)
df
```

```python
# 1.Genre vs Global sales
genre_sales=df.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)
sns.barplot(x=genre_sales.index,y=genre_sales)
plt.title('Total Global Sales for each genre.')
plt.xticks(rotation=45,ha='right')
print(f'the Best selling game overall is --> {genre_sales.idxmax()} <-- Game and least selling game is --> {genre_sales.idxmin()} <--')
plt.show()
```

```python
# 2.Platform vs Sales
topTenPlatform=df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=topTenPlatform.index,y=topTenPlatform,hue=topTenPlatform.index,legend=False)
plt.title('Top 10 Platform With Highest Global Sales.')
print(f'the Best selling Platform is  overall is --> {topTenPlatform.idxmax()} <-- and least selling Platform --> {topTenPlatform.idxmin()} <--')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
# platform that earns the most is "PS2" and Least earn Platform is "PC"
```

```python
# 3.Publisher vs Sales
topTenPublisher=df.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=topTenPublisher.index,y=topTenPublisher,color='SkyBlue')
plt.title('Top 10 Publisher With Highest Global Sales ')
plt.xticks(rotation=45,ha='right')
plt.show()
```

```python
# 4.Year vs Sales
# split years into High (≥100M), Mid (10M–100M), and Low (<10M) groups for better Understanding The Trends
Years_Sales=df.groupby('Year')['Global_Sales'].sum().sort_index()
high_sales = Years_Sales[Years_Sales >= 100]      #--> High sales years
mid_sales  = Years_Sales[(Years_Sales < 100) & (Years_Sales >= 10)]  #-->Mid sales years
low_sales  = Years_Sales[Years_Sales < 10]        #-->Low sales years
plt.figure(figsize=(10,8))

# HIGH SALES YEARS
plt.subplot(3,1,1)
sns.barplot(x=high_sales.index,y=high_sales,color='Green')
plt.title("High Sales Years (≥ 100M)")
plt.xticks(rotation=45, ha='right')
plt.ylabel("Global Sales (millions)")

# MID SALES YEARS
plt.subplot(3,1,2)
sns.barplot(x=mid_sales.index,y=mid_sales,color='Orange')
plt.title("Mid Sales Years (10M–100M)")
plt.xticks(rotation=45, ha='right')
plt.ylabel("Global Sales (millions)")

# LOW SALES YEARS
plt.subplot(3,1,3)
sns.barplot(x=low_sales.index,y=low_sales,color='Red')
plt.title("Low Sales Years (< 10M))")
plt.xticks(rotation=45, ha='right')
plt.ylabel("Global Sales (millions)")

plt.tight_layout()
plt.show()
```
